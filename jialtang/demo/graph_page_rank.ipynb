```python
import networkx as nx

# 1. 创建一个无向图
G = nx.Graph()

# 添加节点和边
edges = [
    ('A', 'B'), ('A', 'C'),
    ('B', 'C'), ('B', 'D'),
    ('C', 'D'),
    ('D', 'E'),
    ('E', 'F'),
    ('F', 'G'),
    ('G', 'H'),
    ('H', 'E')
]

G.add_edges_from(edges)

# 2. 计算 PageRank
pagerank_scores = nx.pagerank(G)

# 3. 打印结果
print("节点 PageRank 分数 (降序排列):")
print("-" * 35)
for node, score in sorted(pagerank_scores.items(), key=lambda x: x[1], reverse=True):
    print(f"  节点 {node}: {score:.6f}")

# 4. 统计信息
print(f"\n图的基本信息:")
print(f"  节点数: {G.number_of_nodes()}")
print(f"  边数: {G.number_of_edges()}")
degrees = dict(G.degree())
print(f"  节点度分布: {degrees}")
print(f"  平均度: {sum(degrees.values()) / G.number_of_nodes():.2f}")
```

```python
# 创建另一个图示例
G2 = nx.Graph()
G2.add_edges_from([
    (1, 2), (1, 3), (1, 4),
    (2, 3), (2, 5),
    (3, 4), (3, 5),
    (4, 6),
    (5, 6), (5, 7),
    (6, 7), (6, 8),
    (7, 8)
])

print("=" * 50)
print("示例 2: 不同参数设置")
print("=" * 50)

# 1. 默认参数
pr_default = nx.pagerank(G2)
print("\n1. 默认参数 (alpha=0.85):")
for node in sorted(G2.nodes()):
    print(f"  节点 {node}: {pr_default[node]:.6f}")

# 2. 调整阻尼因子
pr_alpha_07 = nx.pagerank(G2, alpha=0.7)  # 更多随机跳转
pr_alpha_095 = nx.pagerank(G2, alpha=0.95)  # 更少随机跳转

print("\n2. 不同阻尼因子的比较:")
print("  节点 | alpha=0.70 | alpha=0.85 | alpha=0.95")
print("  " + "-" * 45)
for node in [1, 3, 6, 8]:  # 显示部分节点
    print(f"  {node:2d}   | {pr_alpha_07[node]:.6f} | {pr_default[node]:.6f} | {pr_alpha_095[node]:.6f}")

# 3. 个性化 PageRank
print("\n3. 个性化 PageRank:")
# 假设我们更关注节点1和节点6
personalization = {node: 0.0 for node in G2.nodes()}
personalization[1] = 0.7
personalization[6] = 0.3

pr_personalized = nx.pagerank(G2, personalization=personalization)
print("  个性化向量: 节点1(0.7), 节点6(0.3), 其他节点(0.0)")
for node in G2.nodes():
    if pr_personalized[node] > 0.12:  # 显示较高的节点
        print(f"  节点 {node}: {pr_personalized[node]:.6f}")

# 4. 带权重的 PageRank
print("\n4. 带权重的 PageRank:")
G_weighted = nx.Graph()
# 添加带权重的边
G_weighted.add_weighted_edges_from([
    ('X', 'Y', 2.0), ('X', 'Z', 1.0),
    ('Y', 'Z', 3.0), ('Y', 'W', 0.5),
    ('Z', 'W', 1.5),
    ('W', 'V', 2.0)
])

pr_weighted = nx.pagerank(G_weighted, weight='weight')
for node, score in sorted(pr_weighted.items()):
    print(f"  节点 {node}: {score:.6f}")
```
